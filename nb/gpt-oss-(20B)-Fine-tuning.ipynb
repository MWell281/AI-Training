### Training


```python
import json
import requests
from datasets import Dataset

LANG = "python"
TARGET_ROWS = 240
MAX_LENGTH = 1024

assert "tokenizer" in globals(), "请先运行模型加载格"

url = (
    "https://huggingface.co/datasets/bigcode/commitpackft/"
    f"resolve/main/data/{LANG}/data.jsonl"
)

rows = []
seen = set()
allowed_licenses = {
    "mit", "apache-2.0", "bsd-2-clause", "bsd-3-clause",
    "isc", "unlicense", "cc0-1.0",
}

with requests.get(url, stream=True, timeout=120) as response:
    response.raise_for_status()

    for raw_line in response.iter_lines():
        if not raw_line:
            continue

        item = json.loads(raw_line.decode("utf-8"))

        if (item.get("license") or "").lower() not in allowed_licenses:
            continue

        old_code = item.get("old_contents") or ""
        new_code = item.get("new_contents") or ""
        instruction = (item.get("subject") or "").strip()
        filename = item.get("old_file") or "file.py"

        if not old_code.strip() or not new_code.strip() or not instruction:
            continue
        if old_code == new_code:
            continue

        key = (old_code, new_code)
        if key in seen:
            continue

        user_prompt = (
            f"Modify this existing {LANG} file.\n\n"
            f"File: {filename}\n\n"
            f"Request:\n{instruction}\n\n"
            f"Current file:\n{old_code}\n\n"
            "Return the complete revised file only."
        )

        messages = [
            {"role": "user", "content": user_prompt},
            {"role": "assistant", "content": new_code},
        ]

        text = tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=False,
            reasoning_effort="low",
        )

        # 按完整聊天格式计算长度；超长样本跳过
        token_count = len(
            tokenizer(text, add_special_tokens=False)["input_ids"]
        )
        if token_count > MAX_LENGTH:
            continue

        seen.add(key)
        rows.append({"text": text})

        if len(rows) >= TARGET_ROWS:
            break

print("可用样本：", len(rows))
assert len(rows) >= 30, "短样本不足，暂时不要开始训练"

commit_data = Dataset.from_list(rows).train_test_split(
    test_size=0.1,
    seed=42,
)

print(commit_data)
print(commit_data["train"][0]["text"][:800])
```

```python
from trl import SFTConfig, SFTTrainer

FastLanguageModel.for_training(model)

trainer = SFTTrainer(
    model=model,
    processing_class=tokenizer,
    train_dataset=commit_data["train"],
    eval_dataset=commit_data["test"],
    args=SFTConfig(
        output_dir="/content/commitpack-v2-output",
        dataset_text_field="text",
        max_length=1024,
        per_device_train_batch_size=1,
        per_device_eval_batch_size=1,
        gradient_accumulation_steps=8,
        max_steps=10,
        learning_rate=1e-5,
        logging_steps=1,
        save_strategy="no",
        report_to="none",
    ),
)

trainer.train()
print(trainer.evaluate())
```

```python
from unsloth import FastLanguageModel

FastLanguageModel.for_inference(model)

# 记住当前 v2 adapter
v2_adapter = model.active_adapter

# 加载你保存的 v1；不需要再加载第二个完整模型
model.load_adapter(
    "/content/drive/MyDrive/AI learning/gpt-oss-fable-lora-v1",
    adapter_name="compare_v1",
    is_trainable=False,
)

question = (
    "Fix this Python function:\n\n"
    "def merge_settings(defaults, overrides):\n"
    "    defaults.update(overrides)\n"
    "    return defaults\n\n"
    "Requirements:\n"
    "- Recursively merge nested dictionaries.\n"
    "- Overrides take precedence.\n"
    "- Replace lists rather than concatenate them.\n"
    "- Do not modify either input.\n"
    "- Changes to nested dictionaries or lists in the result "
    "must not affect either input.\n"
    "- Use only the standard library.\n"
    "- Return only the corrected function."
)

def get_answer():
    inputs = tokenizer.apply_chat_template(
        [{"role": "user", "content": question}],
        add_generation_prompt=True,
        return_tensors="pt",
        reasoning_effort="low",
    ).to("cuda")

    output = model.generate(
        inputs,
        max_new_tokens=512,
        do_sample=False,
        pad_token_id=tokenizer.eos_token_id,
    )

    return tokenizer.decode(
        output[0][inputs.shape[-1]:],
        skip_special_tokens=True,
    )

try:
    model.set_adapter("compare_v1")
    v1_answer = get_answer()

    model.set_adapter(v2_adapter)
    v2_answer = get_answer()
finally:
    model.set_adapter(v2_adapter)

print("===== v1 =====")
print(v1_answer)
print("\n===== v2 =====")
print(v2_answer)
```

### Reasoning Effort
The `gpt-oss` models from OpenAI include a feature that allows users to adjust the model's "reasoning effort." This gives you control over the trade-off between the model's performance and its response speed (latency) which by the amount of token the model will use to think.

----

The `gpt-oss` models offer three distinct levels of reasoning effort you can choose from:

* **Low**: Optimized for tasks that need very fast responses and don't require complex, multi-step reasoning.
* **Medium**: A balance between performance and speed.
* **High**: Provides the strongest reasoning performance for tasks that require it, though this results in higher latency.

```python
from transformers import TextStreamer

messages = [
    {"role": "user", "content": "Solve x^5 + 3x^4 - 10 = 3."},
]
inputs = tokenizer.apply_chat_template(
    messages,
    add_generation_prompt = True,
    return_tensors = "pt",
    return_dict = True,
    reasoning_effort = "low", # **NEW!** Set reasoning effort to low, medium or high
).to("cuda")

_ = model.generate(**inputs, max_new_tokens = 64, streamer = TextStreamer(tokenizer))
```

Changing the `reasoning_effort` to `medium` will make the model think longer. We have to increase the `max_new_tokens` to occupy the amount of the generated tokens but it will give better and more correct answer

### Data Prep

The `HuggingFaceH4/Multilingual-Thinking` dataset will be utilized as our example. This dataset, available on Hugging Face, contains reasoning chain-of-thought examples derived from user questions that have been translated from English into four other languages. It is also the same dataset referenced in OpenAI's [cookbook](https://cookbook.openai.com/articles/gpt-oss/fine-tune-transfomers) for fine-tuning. The purpose of using this dataset is to enable the model to learn and develop reasoning capabilities in these four distinct languages.

```python
def formatting_prompts_func(examples):
    convos = examples["messages"]
    texts = [tokenizer.apply_chat_template(convo, tokenize = False, add_generation_prompt = False) for convo in convos]
    return { "text" : texts, }

from datasets import load_dataset

dataset = load_dataset("HuggingFaceH4/Multilingual-Thinking", split = "train")
dataset
```

To format our dataset, we will apply our version of the GPT OSS prompt

```python
from unsloth.chat_templates import standardize_sharegpt
dataset = standardize_sharegpt(dataset)
dataset = dataset.map(formatting_prompts_func, batched = True,)
```

### Train the model
Now let's train our model. We do 60 steps to speed things up, but you can set `num_train_epochs=1` for a full run, and turn off `max_steps=None`.

```python
from trl import SFTConfig, SFTTrainer
trainer = SFTTrainer(
    model = model,
    tokenizer = tokenizer,
    train_dataset = dataset,
    args = SFTConfig(
        per_device_train_batch_size = 1,
        gradient_accumulation_steps = 4,
        warmup_steps = 5,
        # num_train_epochs = 1, # Set this for 1 full training run.
        max_steps = 30,
        learning_rate = 2e-4,
        logging_steps = 1,
        optim = "adamw_8bit",
        weight_decay = 0.001,
        lr_scheduler_type = "linear",
        seed = 3407,
        output_dir = "outputs",
        report_to = "none", # Use TrackIO/WandB etc
    ),
)
```

We also use Unsloth's `train_on_completions` method to only train on the assistant outputs and ignore the loss on the user's inputs. This helps increase accuracy of finetunes and lower loss as well!

```python
from unsloth.chat_templates import train_on_responses_only

gpt_oss_kwargs = dict(instruction_part = "<|start|>user<|message|>", response_part = "<|start|>assistant<|channel|>final<|message|>")

trainer = train_on_responses_only(
    trainer,
    **gpt_oss_kwargs,
)
```
